# Week 8 · Generators produce one result at a time

Define countdown(n) yielding n down to 1; nonpositive n yields nothing.

A generator uses yield to produce a value and pause. The next iteration resumes from that point. It avoids creating a full list when you only need one item at a time.

Generators are consumed as you iterate. To repeat the sequence, create a new generator. They are useful for processing a stream or large file without keeping every result in memory.
